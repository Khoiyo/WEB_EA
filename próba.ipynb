```python
print("1.")
szoveg = text.replace('?', '.').replace('!', '.')
nyers = szoveg.split('.')

for mondat in nyers:
    m = mondat.strip()
    if len(m) > 0:
        print(m + ".")

print("\n2.")
fajl = open("szoveg.txt", "w", encoding="utf-8")
fajl.write(text)
fajl.close()
print("Fájlba írás kész.")

print("\n3.")
tiszta = text.lower()

irasjelek = [".", ",", "!", "?", ":", ";", '"', "'"]
for jel in irasjelek:
    tiszta = tiszta.replace(jel, "")

szavak = tiszta.split()

szotar = {}
for szo in szavak:
    if szo in szotar:
        szotar[szo] += 1
    else:
        szotar[szo] = 1

print(szotar)

print("\n4.")
leghosszabb = ""
for szo in szotar:
    if len(szo) > len(leghosszabb):
        leghosszabb = szo

print(f"A leghosszabb szó: {leghosszabb} ({len(leghosszabb)} )")

print("\n5.")
egyedi = []
for szo in szotar:
    hossz = len(szo)
    if hossz not in egyedi:
        egyedi.append(hossz)

egyedi.sort()
print("Előforduló szóhosszok:", egyedi)

rendezett = []
for szo in szotar:
    rendezett.append(szo)

for i in range(len(rendezett) - 1):
    for j in range(i + 1, len(rendezett)):
        if len(rendezett[i]) > len(rendezett[j]):
            temp = rendezett[i]
            rendezett[i] = rendezett[j]
            rendezett[j] = temp

print("\nhossz szerint rendezve:")
for szo in rendezett:
    print(f"{szo}: {len(szo)}")
```

```python
import math

def prim(n):
    if n < 2:
        return False
    for i in range(2, int(math.sqrt(n)) + 1):
        if n % i == 0:
            return False
    return True

def kiir(hatar):
    if not isinstance(hatar, int) or not (1000 <= hatar <= 2000):
        print("Hiba!")
        return

    fajlnev = "primek.csv"
    try:
        with open(fajlnev, "w", encoding="utf-8") as fajl:
            fajl.write("Sorszam;Prím\n")

            sorszam = 1
            for szam in range(2, hatar + 1):
                if prim(szam):
                    fajl.write(f"{sorszam};{szam}\n")
                    sorszam += 1

        print(f"Sikeres kiírás!")

    except IOError:
        print("Hiba!")

kiir(1500)
```
